```python
# Major Toyota models from the Integrated Report
major_models = [
    "Corolla",
    "RAV4",
    "Yaris",
    "Camry",
    "Hilux",
    "Highlander",
    "Sienna",
    "Tacoma",
    "Lexus RX",
    "Innova"
]

# Map report names to names used in the cleaned specs dataset
model_mapping = {
    "Highlander": "Highlander / Kluger",
    "RAV4": "RAV4",
}

major_models_specs = []

for report_model in major_models:
    spec_model = model_mapping.get(report_model, report_model)

    subset = toyota_clean[
        toyota_clean["model"] == spec_model
    ]

    major_models_specs.append(
        subset[["model", "powertrain_category"]]
        .drop_duplicates()
        .assign(Report_Model=report_model)
    )

major_models_specs = pd.concat(
    major_models_specs,
    ignore_index=True
)

display(
    major_models_specs[
        ["Report_Model", "model", "powertrain_category"]
    ].sort_values(
        ["Report_Model", "powertrain_category"]
    )
)
```

```python
model_powertrain_summary = (
    major_models_specs
    .groupby("Report_Model")["powertrain_category"]
    .apply(lambda x: ",".join(sorted(x.unique())))
    .reset_index()
)

display(model_powertrain_summary)
```
